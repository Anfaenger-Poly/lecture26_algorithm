### 1. 이진 트리 (Binary Tree)

```python
class TreeNode:
    def __init__(self, data, left, right):
        self.data = data # 멤버
        self.left = left # 멤버
        self.right = right #멤버
   # def set_left(self, left):
        #self.left = left
    #def set_right(self, right):
        #self.right = right
```

```python
# binary tree 생성
d = TreeNode('D', None, None)
e = TreeNode('E', None, None)
b = TreeNode('B', d, e)

f = TreeNode('F', None, None)
c = TreeNode('C', f, None)

a = TreeNode('A', b, c)

my_binary_tree = a
```

### preorder (전위 순회)

```python
def preorder(node):
    #if node == None: return

    print(node.data, end =' ')
    if node.left : preorder(node.left)
    if node.right : preorder(node.right)

preorder(my_binary_tree)
```

### inorder (중위 순회)

```python
def inorder(node):
    if node == None: return

    inorder(node.left)
    print(node.data, end =' ')
    inorder(node.right)

inorder(my_binary_tree)
```

### postorder(후위 순회)

```python
def postorder(node):
    if node == None: return

    postorder(node.left)
    postorder(node.right)
    print(node.data, end =' ')

postorder(my_binary_tree)
```

### Binary Search Tree

```python
class StreeNode:
    def __init__(self, key, left, right):
        self.key = key # 멤버
        self.left = left # 멤버
        self.right = right
```

```python
# BST 생성
t3 = StreeNode(3, None, None)
t12 = StreeNode(12, None, None)
t7 = StreeNode(7, t3, t12)
t27 = StreeNode(27, None, None)
t31 = StreeNode(31, t27, None)
t26 = StreeNode(26, None, t31)
t18 = StreeNode(18, t7, t26)
my_bst = t18
```

```python
def search_bst(n, key):
    if n == None: return None
    if key == n.key: return n
    elif key < n.key:
        return search_bst(n.left, key)
    else:
        return search_bst(n.right, key)

result = search_bst(my_bst, 12)
if result: print(result.key)
else:
    print('없는 key입니다')
```

```python
def search_bst_iterative(n, key):
    while n != None:
        if key == n.key: return n
        elif key < n.key:
            n = n.left
        else:
            n = n.right
    return None

result = search_bst_iterative(my_bst, 31)
if result: print(result.key)
else:
    print('없는 key입니다')
```

### 숫자를 순서대로 입역받아 Binary Search Tree 구성

```python
class STreeNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

def insert_bst(root, key):
    new_node = STreeNode(key)
    
    # 트리가 비어있으면 새 노드가 root가 됨
    if root is None:
        return new_node
    
    current = root
    parent = None

    while current is not None:
        parent = current
        if key < current.key:
            current = current.left # 작으면 왼쪽
        elif key > current.key:
            current = current.right # 크면 오른쪽
        else:
            return root  # 중복된 키는 삽입하지 않음
        
    # 부모 노드의 어느 쪽에 연결할지 결정
    if key < parent.key:
        parent.left = new_node
    else:
        parent.right = new_node
    return root

# BST 생성
my_bst = None
numbers = [2, 32, 11, 27, 20, 18, 6, 13, 31]

for num in numbers:
    my_bst = insert_bst(my_bst, num)

# 기존 search 함수로 검증
def inorder(n):
    if n is None:
        return
    inorder(n.left)
    print(n.key, end=' ')
    inorder(n.right)

inorder(my_bst)  # 삽입된 노드들이 올바르게 배치되었는지 확인
        
```
